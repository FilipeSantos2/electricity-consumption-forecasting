## When Does MT_001 Start Recording Consumption?

```python
nonzero = df["MT_001"] > 0

print("First nonzero measurement:")
print(df.loc[nonzero, "Unnamed: 0"].iloc[0])

print("Percentage of zero measurements:")
print((df["MT_001"] == 0).mean() * 100)
```

```python
df = df.rename(columns={"Unnamed: 0": "timestamp"})

df["timestamp"] = pd.to_datetime(
    df["timestamp"],
    format="%Y-%m-%d %H:%M:%S"
)
```

```python
print(df["timestamp"].dtype)
print(df["timestamp"].head())
```

# Plot a Week in Which the Consumer was Active

```python
first_nonzero = df.loc[
    df["MT_001"] > 0,
    "timestamp"
].iloc[0]

print(first_nonzero)
```

```python
week = df[
    (df["timestamp"] >= first_nonzero) &
    (df["timestamp"] < first_nonzero + pd.Timedelta(days=7))
]
```

```python
plt.figure(figsize=(14, 5))

plt.plot(week["timestamp"], week["MT_001"])

plt.title("Electricity Consumption - MT_001 (First Active Week)")
plt.xlabel("Timestamp")
plt.ylabel("Power (kW)")
plt.xticks(rotation=45)
plt.grid()
plt.tight_layout()
plt.show()
```

# Quantifying the Zeros

```python
before_2012 = df["timestamp"] < "2012-01-01"
from_2012 = ~before_2012

print("Measurements before 2012:")
print("Total:", before_2012.sum())
print("Zeros:", (df.loc[before_2012, "MT_001"] == 0).sum())

print("\nMeasurements from 2012 onwards:")
print("Total:", from_2012.sum())
print("Zeros:", (df.loc[from_2012, "MT_001"] == 0).sum())
print(
    "Percentage of zeros:",
    (df.loc[from_2012, "MT_001"] == 0).mean() * 100
)
```

# January 2012 Plot for MT_001

```python
january = df[
    (df["timestamp"] >= "2012-01-01") &
    (df["timestamp"] < "2012-02-01")
]

plt.figure(figsize=(15, 5))

plt.plot(
    january["timestamp"],
    january["MT_001"],
    linewidth=0.8
)

plt.title("Electricity Consumption - MT_001 (January 2012)")
plt.xlabel("Timestamp")
plt.ylabel("Power (kW)")
plt.xticks(rotation=45)
plt.grid()
plt.tight_layout()
plt.show()
```
